### Code 2-1: Enumerating and Evaluating 3-Bit States

This example generates all possible 3-bit combinations and evaluates each one. For every state, we check whether exactly two bits are set to 1. The key idea is that each configuration is handled independently. The program works through the full set of possibilities step by step, making the process clear and predictable.

```python
from itertools import product

N = 3  # number of bits (try changing this)

def evaluate(state):
    return sum(state) == 2

# generate all 3-bit combinations
states = list(product([0, 1], repeat=N))

# evaluate each state
for s in states:
    print(f"{s} -> {evaluate(s)}")
```

### Code 2-2: A First Quantum Circuit

This example introduces a minimal quantum program using Qiskit. We define a circuit with a single qubit and a classical bit, apply a Hadamard gate to place the qubit into superposition, and then measure it. The diagram shows the flow from preparation to transformation to measurement, giving a visual sense of how quantum programs are structured.

**Note:** Be sure to run the required `pip install` step for Qiskit before executing this cell.

```python
!pip -q install qiskit qiskit_aer pylatexenc
```

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(1, 1)  # 1 qubit, 1 classical bit (measurement storage)
qc.h(0)                    # put qubit into superposition
qc.measure(0, 0)           # measure qubit -> classical bit

qc.draw('mpl')             # visualize the circuit
```

### Code 2-3: Running the Circuit Multiple Times

Here we execute the same circuit repeatedly using a simulator. Each run produces either 0 or 1, but over many runs the results are roughly evenly split. This shows that the qubit is not fixed to a single value after the Hadamard gate, but instead produces outcomes based on probabilities that emerge when the system is measured.

**Note:** This cell depends on `qc` from the prior cell being defined and executed first.

```python
from qiskit_aer import AerSimulator

# Assumes qc was created in the previous cell

shots = 10_000
sim = AerSimulator()
result = sim.run(qc, shots=shots).result()
counts = result.get_counts()

print(f"Results over {shots:,} runs: {counts}")
```

### Code 2-4: Three-Qubit Superposition and Measurement

This example extends the earlier single-qubit circuit to three qubits. Each qubit is placed into superposition using a Hadamard gate, allowing the system to represent all eight possible states. The circuit is then executed multiple times, showing how measurement produces one outcome per run while revealing a distribution across all possibilities.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from IPython.display import display

qc = QuantumCircuit(3, 3)      # 3 qubits, 3 classical bits
qc.h([0, 1, 2])                # put all qubits into superposition
qc.measure([0, 1, 2], [0, 1, 2])

display(qc.draw('mpl'))        # force the circuit diagram to display

shots = 10_000
sim = AerSimulator()
counts = sim.run(qc, shots=shots).result().get_counts()

print(f"Results over {shots:,} runs: {counts}")
```

### Code 2-5: Biasing Outcomes with Amplitude

This circuit prepares a single qubit using an Ry rotation instead of a Hadamard gate. The rotation changes how much each possible outcome contributes to the overall state. When measured repeatedly, the results are no longer evenly split. One outcome appears more often, reflecting the underlying amplitudes. This shows how small changes in state preparation directly shape the distribution of observed results.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from IPython.display import display

qc = QuantumCircuit(1, 1)  # Create circuit with 1 qubit and 1 classical bit

# Prepare a biased superposition (amplitude)
qc.ry(1.0, 0)              # rotate around Y-axis (angle in radians)

qc.measure(0, 0)           # Measure

# Display circuit
display(qc.draw('mpl'))

# Run simulation
shots = 10_000
sim = AerSimulator()
result = sim.run(qc, shots=shots).result()
counts = result.get_counts()

print(f"Results over {shots:,} runs: {counts}")
```

### Code 2-6: Interference Cancels One Outcome

This circuit applies two Hadamard gates in sequence to a single qubit. The first creates a balanced superposition, while the second causes the contributions to interact. Through interference, the paths leading to one outcome cancel, while the other is reinforced. When measured repeatedly, the result is no longer distributed. Instead, the circuit consistently produces a single deterministic outcome.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from IPython.display import display

qc = QuantumCircuit(1, 1)  # 1 qubit, 1 classical bit

# Create superposition
qc.h(0)

# Apply a second transformation
qc.h(0)

# Measure
qc.measure(0, 0)

# Display circuit
display(qc.draw('mpl'))

# Run simulation
shots = 10_000
sim = AerSimulator()
result = sim.run(qc, shots=shots).result()
counts = result.get_counts()

print(f"Results over {shots:,} runs: {counts}")
```

### Code 2-7: Creating an Entangled Pair

This example builds a simple two-qubit circuit that demonstrates entanglement. The `QuantumCircuit` is initialized with two qubits, a Hadamard gate `h` places the first qubit into superposition, and a controlled-NOT `cx` links the second qubit to it. After `measure`, the simulator runs the circuit many times, showing correlated outcomes that reveal how entangled qubits behave together.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from IPython.display import display

qc = QuantumCircuit(2, 2)  # 2 qubits, 2 classical bits

qc.h(0)                    # put qubit 0 into superposition
qc.cx(0, 1)                # entangle qubit 1 with qubit 0

qc.measure([0, 1], [0, 1]) # measure both qubits

display(qc.draw('mpl'))    # display circuit

shots = 10_000
sim = AerSimulator()
result = sim.run(qc, shots=shots).result()
counts = result.get_counts()

print(f"Results over {shots:,} runs: {counts}")
```

### Code 2-8: Visualizing Qubit Rotation on the Bloch Sphere

This example visualizes how a single-qubit rotation changes the state of a qubit using a Bloch sphere. The point at the top represents the initial state |0⟩. After applying an Ry rotation, the state moves along a smooth arc to a new position on the sphere. This new point reflects a different balance between 0 and 1. The figure connects the code directly to the physical interpretation, showing how gates shape the state before measurement.

```python
import numpy as np
import matplotlib.pyplot as plt

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

theta = 1.0

qc = QuantumCircuit(1)
qc.ry(theta, 0)

state = Statevector.from_instruction(qc)
alpha, beta = state.data

# Bloch coordinates
x = 2 * np.real(np.conj(alpha) * beta)
y = 2 * np.imag(np.conj(alpha) * beta)
z = abs(alpha)**2 - abs(beta)**2

# --- Plot ---
fig = plt.figure(figsize=(12, 12))
fig.patch.set_facecolor('white')
ax = fig.add_subplot(111, projection="3d")

# --- REMOVE ALL AXES STRUCTURE ---
ax.set_axis_off()  # <- this is the key line

# --- Sphere ---
u = np.linspace(0, 2*np.pi, 40)
v = np.linspace(0, np.pi, 20)
xs = np.outer(np.cos(u), np.sin(v))
ys = np.outer(np.sin(u), np.sin(v))
zs = np.outer(np.ones_like(u), np.cos(v))
ax.plot_wireframe(xs, ys, zs, color='gray', alpha=0.2)

# --- Start |0> ---
ax.scatter(0, 0, 1, s=180)

# shift slightly right (+x) and up (+z)
ax.text(0.08, 0, 1.2, "|0⟩",
        ha='left', fontsize=15)


# --- Final state ---
ax.scatter(x, y, z, s=200)

# shift slightly right (+x) and up (+z)
ax.text(x + 0.08, y, z + 0.12, "After Ry(1.0)",
        ha='left', fontsize=14)

# --- Rotation arc ---
t = np.linspace(0, theta, 100)
arc_x = np.sin(t)
arc_y = np.zeros_like(t)
arc_z = np.cos(t)
ax.plot(arc_x, arc_y, arc_z, linewidth=2.5)

# Optional label for arc
ax.text(.6, .3, 0.75, "Rotation", ha='center', fontsize=12)

# Keep sphere proportions
ax.set_box_aspect([1, 1, 1])
ax.set_xlim([-1, 1])
ax.set_ylim([-1, 1])
ax.set_zlim([-1, 1])

# Title
ax.set_title("Qubit Rotation from Code to State", fontsize=16, pad=20)

plt.show()
```
