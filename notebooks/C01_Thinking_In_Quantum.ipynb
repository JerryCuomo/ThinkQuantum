### Utility Program to generate Chapter 1's Figure 1-8 timeline

This notebook provides a supporting role for Chapter 1. Rather than focusing on programming, it is used to generate the timeline figure that appears in the chapter.

The code uses a plotting utility to construct a log-scale view of qubit growth and position key milestones such as `Quantum Supremacy`, `Fault Tolerance`, `Quantum Advantage`, and the `Q-Day Threshold`. Visual elements, including the duck figures, are added to make each stage easier to recognize and compare.

You can run the notebook to reproduce the figure and make adjustments to labels, scaling, or annotations as needed. This is a simple example of using code to organize and present technical ideas in a visual format...

The next chapter introduces how to write and run quantum programs.

```python
import matplotlib.pyplot as plt

# -----------------------------
# Style Constants (easy tuning)
# -----------------------------
TITLE_FONTSIZE = 14
DESC_FONTSIZE = 12
TICK_FONTSIZE = 12
SUBTITLE_FONTSIZE = 11

COLOR_TITLE = "0.15"
COLOR_DESC = "0.20"
COLOR_TICKS = "0.20"
COLOR_SUBTITLE = "0.20"
COLOR_AXIS = "0.35"
COLOR_GUIDE = "0.35"

# -----------------------------
# Data
# -----------------------------
events = [
    {"x": 1e2, "title": "Quantum Supremacy", "desc": "Narrow task beyond\nclassical reach"},
    {"x": 1e3, "title": "Early Utility", "desc": "First exploration of\nreal problems"},
    {"x": 1e4, "title": "Fault Tolerance", "desc": "Emergence of\nlogical qubits"},
    {"x": 1e5, "title": "Quantum Advantage", "desc": "Practical benefit\nover classical"},
    {"x": 1e6, "title": "Q-Day Threshold", "desc": "Classical encryption\nbecomes vulnerable"},
    {"x": 1e8, "title": "Transformational Scale", "desc": "Broad impact across\nscience and industry"},
]

# -----------------------------
# Figure setup
# -----------------------------
fig, ax = plt.subplots(figsize=(14, 5))

# Main timeline
ax.hlines(0, 1e2, 1e8, color=COLOR_AXIS, linewidth=1.2)

# Alternating label heights
label_heights = [0.75, 1.55, 0.75, 1.55, 0.75, 1.55]

# -----------------------------
# Plot events
# -----------------------------
for event, y in zip(events, label_heights):
    x = event["x"]

    # Vertical guide
    ax.vlines(x, 0, y - 0.08, color=COLOR_GUIDE, linewidth=1.0)

    # Tick marker
    ax.plot(x, 0, marker="o", markersize=4, color=COLOR_AXIS)

    # Title
    ax.text(
        x, y,
        event["title"],
        ha="center",
        va="bottom",
        fontsize=TITLE_FONTSIZE,
        fontweight="bold",
        color=COLOR_TITLE
    )

    # Description
    ax.text(
        x, y - 0.12,
        event["desc"],
        ha="center",
        va="top",
        fontsize=DESC_FONTSIZE,
        color=COLOR_DESC
    )

# -----------------------------
# Axis setup
# -----------------------------
ax.set_xscale("log")
ax.set_xlim(7e1, 2e8)
ax.set_ylim(-0.28, 1.35)

ticks = [1e2, 1e3, 1e4, 1e5, 1e6, 1e8]

tick_labels = [
    "$10^2$\n100 qubits",
    "$10^3$\n1,000 qubits",
    "$10^4$\n10,000 qubits",
    "$10^5$\n100,000 qubits",
    "$10^6$\n1M qubits",
    "$10^8$\n100+M qubits",
]

ax.set_xticks(ticks)
ax.set_xticklabels(tick_labels, fontsize=TICK_FONTSIZE, color=COLOR_TICKS)

# Subtitle
ax.text(
    1.05e2, -0.20,
    "Physical qubits (log scale)",
    fontsize=SUBTITLE_FONTSIZE,
    color=COLOR_SUBTITLE,
    ha="left",
    va="top"
)

# -----------------------------
# Clean styling
# -----------------------------
ax.set_yticks([])

for spine in ["left", "right", "top", "bottom"]:
    ax.spines[spine].set_visible(False)

ax.tick_params(axis="x", length=0)

plt.tight_layout()
plt.show()
```
